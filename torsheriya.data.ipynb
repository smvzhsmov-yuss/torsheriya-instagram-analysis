```python
import pandas as pd
from pathlib import Path

file_path = Path.home() / "Desktop" / "Ig_Project_data.xlsx"
df = pd.read_excel(file_path)
df.head()
```

```python
pd.set_option("display.max_rows", None)
display(df)
print("Total records:", len(df))
```

```python
print(df.columns.tolist())
```

```python
# of total interactions
df["engagement"] = (
    df["likes"].fillna(0) + df["comments"].fillna(0) + df["shares"].fillna(0) + df["saves"].fillna(0))
```

```python
display(df[["post_Id", "theme", "engagement"]])
```

```python
display(df[["post_Id", "time", "engagement"]])
```

```python
results = df.groupby("time").agg(
    average_engagement=("engagement", "mean"),
    number_of_posts=("engagement", "count")
)
```

```python
df.nlargest(43, "engagement")[["post_Id", "theme", "time", "engagement"]]
```

```python
df["time"]= pd.to_datetime(df["time"], format="%I:%M%p", errors="coerce").dt.hour
```

```python
df.groupby("time")["engagement"].mean().plot(
    kind="line",
    marker="o",
    color="pink",
    title="Best Time To Post",
    xlabel="Hour of Day",
    ylabel="Average Engagement",
    xticks=range(8, 23),
    figsize=(10, 5)
)
```

```python

df['engagement_rate'] = (
    (df['likes'] + df['comments'] + df['shares'])
    / df['views'] * 100
)
```

```python

import matplotlib.pyplot as plt

top10 = df.nlargest(10, 'engagement_rate')

plt.figure(figsize=(10, 5))
plt.bar(top10['post_Id'], top10['engagement_rate'], color='pink')

plt.xlabel('Posts')
plt.ylabel('Engagement Rate (%)')
plt.title('Top 10 Posts by Engagement Rate')
plt.xticks(rotation=45)
plt.show()
```
